```python
# ========================================
# Nama  : Hispano Suiza Artileri
# Tim   : TITAN (Polines)
# Tugas : Menentukan kategori BMI
# ========================================

# Program Kategori BMI
# Input: berat badan (kg) dan tinggi badan (m)

berat = float(input("Masukkan berat badan (kg): "))
tinggi = float(input("Masukkan tinggi badan (m): "))

# Hitung BMI
bmi = berat / (tinggi ** 2)
print(f"\nBMI Anda: {bmi:.2f}")

# Tentukan kategori BMI
if bmi < 18.5:
    kategori = "Kurus"
elif bmi >= 18.5 and bmi < 25:
    kategori = "Normal"
elif bmi >= 25 and bmi < 30:
    kategori = "Kelebihan berat badan"
else:
    kategori = "Obesitas"

# Nested condition untuk tingkat obesitas
if kategori == "Obesitas":
    if bmi >= 40:
        tingkat = "Obesitas Kelas III (Sangat Parah)"
    elif bmi >= 35:
        tingkat = "Obesitas Kelas II (Parah)"
    else:
        tingkat = "Obesitas Kelas I"
    print(f"Kategori: {kategori} - {tingkat}")
else:
    print(f"Kategori: {kategori}")


```
